```python
data = pd.read_csv('https://raw.githubusercontent.com/HoGentTIN/dsai-labs/main/data/airline%20ticket%20data.csv', delimiter = ";", parse_dates=['Month']).set_index(['Month'])
data.head()
```

```python
data.plot(y='Tickets', figsize=[10,5])
```

```python
train_data = data[:'2020-12-31']
test_data = data['2021-01-01':]

# train_data.tail()
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing

train_data_tes = ExponentialSmoothing(train_data, trend='add', seasonal='add', seasonal_periods=12, freq='MS').fit()
train_data_tes_fc = train_data_tes.forecast(12)

train_data_tes_fc.head(5)
# wounded_data['TES'] = train_data_tes.level
# # Notice there is NaN for 2024 in column TES!
# wounded_data.tail(24)
```

```python
train_data['Tickets'].plot(legend=True, label='Train')
test_data['Tickets'].plot(legend=True, label='Test')
train_data_tes_fc.plot(marker='.', legend=True, label='Forecast TES')

# test_data.plot(y='Tickets', figsize=[10,5])
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error

y_true = test_data.values
y_pred = train_data_tes_fc.values

mean_absolute_error(y_true, y_pred)
```
